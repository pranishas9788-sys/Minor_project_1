```python
import numpy as np
name = input("Enter student name:")
subjects = ("Python", "Maths", "English", "Science", "Computer")
marks = []

print("\nEnter marks for 5 subjects")

for subject in subjects:
    mark = float(input("Enter " + subject + " mark: "))
    marks.append(mark)

unique_subjects = set(subjects)

student = {
    "Name": name,
    "Subjects": subjects,
    "Marks": marks
}

def calculate_result(marks):
    total = sum(marks)
    average = np.mean(marks)
    highest = np.max(marks)
    lowest = np.min(marks)

    return total, average, highest, lowest
total, average, highest, lowest = calculate_result(marks)

if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

if average >= 40:
    result = "PASS"
else:
    result = "FAIL"


print("\n========== STUDENT RESULT ==========")
print("Name:", student["Name"])

print("\nMarks:")
for i in range(len(subjects)):
    print(subjects[i], ":", marks[i])

print("\nTotal Marks:", total)
print("Average:", round(average, 2))
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Grade:", grade)
print("Result:", result)

print("\nNumber of Unique Subjects:", len(unique_subjects))

with open("student_result.txt", "w") as file:
    file.write("========== STUDENT RESULT ==========\n")
    file.write("Name: " + name + "\n")

    for i in range(len(subjects)):
        file.write(subjects[i] + ": " + str(marks[i]) + "\n")

    file.write("Total: " + str(total) + "\n")
    file.write("Average: " + str(round(average, 2)) + "\n")
    file.write("Highest: " + str(highest) + "\n")
    file.write("Lowest: " + str(lowest) + "\n")
    file.write("Grade: " + grade + "\n")
    file.write("Result: " + result + "\n")

print("\nResult saved successfully in 'student_result'")
```
